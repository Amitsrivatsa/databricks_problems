First Non-Repeating Element in PySpark
1. Problem Statement

Given a large dataset of elements, find the first element that appears exactly once in the dataset.

The first non-repeating element is the element whose frequency is 1 and whose position is the earliest among all elements that occur only once.

For large datasets, implement the solution using PySpark's distributed DataFrame operations without collecting the entire dataset into driver memory.


```python
from pyspark.sql import *
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
```

```python
nums = [4, 5, 1, 2, 4, 1, 5, 3]

def firstnonrepeatingnumber(nums):
    frequency = {}
    
    for n in nums:
        frequency[n] = frequency.get(n, 0) + 1
    
    for n in nums:
        if frequency[n] == 1:
            return n,frequency
        
```

```python
nums = [4, 5, 1, 2, 4, 1, 5, 3]
retunval,freq =  firstnonrepeatingnumber(nums)
print(retunval)
print(freq)
```
