Intersection of Two Datasets
Problem statement

You are given two datasets containing product IDs. Find the product IDs that exist in both datasets.


```python
from pyspark.sql import *
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
from math import *
import builtins
```

```python
products_a = [101, 102, 103, 104, 105]
products_b = [103, 104, 105, 106, 107]

common_ids = list(set(products_a) & set(products_b))

print(common_ids)
```

PySpark — large datasets

If you only need the common rows/IDs, Spark has a built-in intersect() operation.

```python
common_ids = (
    df_a
    .select("product_id")
    .intersect(
        df_b.select("product_id")
    )
)

common_ids.show()
```
