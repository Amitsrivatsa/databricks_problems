Two Sum — Problem Description

You are given an array of integers nums and an integer target.

Your task is to find two different elements in the array whose values add up to the target.

Return the indices of those two elements.


```python
# Example large DataFrame
df = spark.createDataFrame(
    [
        (1, 2),
        (2, 7),
        (3, 11),
        (4, 15),
    ],
    ["id", "value"]
)

target = 9

```

```python
left_df = df.select("id","value").withColumn("targetval", target - col("value"))
display(left_df)

right_df = df.select(col("id").alias("id2"),col("value").alias("value2"))
display(right_df)

joined_df = left_df.alias("a").join(right_df.alias("b"), (left_df.targetval  == right_df.value2 ) & (left_df.id != right_df.id2), "inner").select("a.id","b.id2","a.value","b.value2").filter(col("id") <= col("id2"))


joined_df = joined_df.agg(collect_list(struct("id","id2"))).alias("collection")
display(joined_df)
```
