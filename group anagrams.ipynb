Group Anagrams — Problem Statement

You are given a large collection of strings. Your task is to group together strings that are anagrams of each other.

Two strings are anagrams if they contain the same characters with the same frequencies, but potentially in a different order.


```python
from pyspark.sql import *
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
from math import *
import builtins
```

```python
words = ["eat", "tea", "tan", "ate", "nat"]
```

```python
data = [(word,) for word in words]
schema = ["word"]
```

```python
df = spark.createDataFrame(data,schema)
df.show()
```

```python
anagram_df = df.withColumn("sorted_array", array_sort(split(col("word"), ""))).groupBy(col("sorted_array")).agg(collect_list(col("word")).alias("word"))
anagram_df.show(10)
```

```python
anagram_df = anagram_df.agg(collect_list(col("word")))
display(anagram_df)
```
