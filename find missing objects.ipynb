Missing ID — Large-Scale PySpark
Problem statement

You are given a dataset containing IDs that should contain every integer from 1 to N exactly once, but one ID is missing.

Your task is to find the missing ID.

```python
def find_missing_id(ids, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(ids)

    return expected_sum - actual_sum
```

```python
ids = [1, 2, 3, 5]

print(find_missing_id(ids, 5))
```
