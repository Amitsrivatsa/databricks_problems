Problem statement

Given a large list of values, return all values that appear more than once.


```python
from pyspark.sql import *
from pyspark.sql.functions import *
from pyspark.sql.window import *
from pyspark.sql.types import *
from math import *
import builtins
```

```python
nums = [1, 2, 3, 2, 4, 5, 1, 6, 3, 7]
```

```python
def filterduplicates(nums):
    frequency = {}
    set_dups = set()
    return_dict = {}

    for num in nums:
        frequency[num] = frequency.get(num,0) + 1
    
    max_count = builtins.max(frequency.values())

    for key,value in frequency.items():
        if value == max_count:
            return_dict[key] = value
    
    return return_dict
```

```python
dups = filterduplicates(nums)
print([nums for nums in dups.keys()])
```
