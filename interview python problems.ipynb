## Group Anagrams

You are given a list of strings.

Group together all strings that are anagrams of each other.

```python
words = [
    "listen", "silent", "enlist", "inlets",
    "evil", "vile", "veil", "live",
    "dusty", "study",
    "night", "thing",
    "save", "vase",
    "state", "taste", "teats",
    "angel", "glean", "angle",
    "rat", "tar", "art",
    "below", "elbow",
    "inch", "chin",
    "brag", "grab",
    "cat", "act", "tac",
    "python",
    "spark",
    "databricks",
    "",
    "",
    "a",
    "a",
    "abc", "cab", "bca",
    "foo", "oof",
    "bar"
]



def group_anagrams(words):

    sorted_dic = {}
 
    for val in words:
        sortedkey = "".join(sorted(val))
        if sortedkey not in sorted_dic:
            sorted_dic[sortedkey] = [] 
        
        sorted_dic[sortedkey].append(val)

    return list(sorted_dic.values())

print(group_anagrams(words))



        
```

## Top K Frequent Elements

Given an integer list nums and an integer k, return the k most frequent elements.

```python
import heapq

nums = [
    5, 3, 5, 2, 8, 5, 3, 3, 9, 1,
    2, 2, 2, 7, 8, 8, 8, 8, 4, 6,
    3, 3, 5, 5, 5, 10, 10, 10, 1, 1,
    1, 1, 7, 7, 7, 6, 6, 4, 4, 4,
    4, 4, 9, 9, 9, 9, 11, 11, 11, 12,
    5, 8, 3, 2, 8, 5, 3, 8, 2, 5,
    13, 13, 13, 13, 13, 13, 14, 14, 15, 15,
    15, 16, 17, 17, 17, 17, 18, 18, 18, 19,
    20, 20, 20, 20, 20, 20, 20
]

k = 4

def top_k_frequent(nums, k):
    count_dic = {}

    for num in nums:
        count_dic[num] = count_dic.get(num,0) + 1

    heap = []

    for num, count in count_dic.items():
        heapq.heappush(heap, (count, num))

        if len(heap) > k:
            heapq.heappop(heap)

    return [num for count, num in heap]

# print(count_dic)  # commented out

print(top_k_frequent(nums, k))
```

## Longest Consecutive Sequence

Given an unsorted list of integers, return the length of the longest sequence of consecutive integers.

The numbers do not have to appear next to each other in the input.

```python
nums = [
    44, 122, -10, 7, 129, 88, 3, 124,
    1000, 51, 121, -3, 132, 75, 5, 127,
    90, 120, 200, 8, 126, -8, 131, 55,
    4, 123, 300, 128, -2, 6, 125, 89,
    133, -9, 500, 130, 2, 76, 134, 52,
    1, 91, 400, -7, 54, 53, 126, 122,
    700, 43, -1, 0
]

def longest_consecutive(nums):
    set_nums = set(nums)
    longest = 0
    current = 0
    length = 0
    for num in nums:
        if num - 1  not in set_nums:
            current = num
            length = 1

        while current+1 in set_nums:
            current = current+1
            length = length+1

        longest = max(longest,length)
    
    return longest

print(longest_consecutive(nums))
```

## Isomorphic Strings

Two strings s and t are isomorphic if every character in s can be mapped to exactly one character in t, while preserving order.


```python
s = "paper"
t = "title"

def is_isomorphic(s, t):
    
    mapst = {}
    mapts = {}

    for a,b in zip(s,t):

        if a in mapst and mapst[a] != b:
            return False
        if b in mapts and mapts[b] != a:
            return False

        mapst[a] = b
        mapts[b] = a
    return True

print(is_isomorphic(s, t))
```

## Happy Number

Given a positive integer n, repeatedly replace it with the sum of the squares of its digits.

```python
def happynum(n):
    seen = set()

    while n != 1:
        print(n)
        if n in seen:
            return False

        seen.add(n)

        total = 0

        for digit in str(n):
            total += int(digit) ** 2

        n = total

    return True


print(happynum(19))  # True
print(happynum(2))   # False

    
```
